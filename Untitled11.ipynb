```python
bad_coords = []

for split in ["train", "valid"]:
    for file in (BASE / split / "labels").glob("*.txt"):
        for line_no, line in enumerate(file.read_text().splitlines(), 1):
            parts = line.split()

            if not parts:
                continue

            try:
                coords = [float(x) for x in parts[1:]]

                for value in coords:
                    if value < 0 or value > 1:
                        bad_coords.append(
                            (split, file.name, line_no, value)
                        )

            except ValueError:
                bad_coords.append(
                    (split, file.name, line_no, "NON_NUMERIC")
                )

print("Bad coordinate values:", len(bad_coords))

for item in bad_coords[:10]:
    print(item)
```

```python
suspicious = []

for split in ["train", "valid"]:
    for file in (BASE / split / "labels").glob("*.txt"):
        for line_no, line in enumerate(file.read_text().splitlines(), 1):
            parts = line.split()

            if len(parts) < 7:
                continue

            coords = [float(x) for x in parts[1:]]

            xs = coords[0::2]
            ys = coords[1::2]

            width = max(xs) - min(xs)
            height = max(ys) - min(ys)

            if width < 0.005 or height < 0.005:
                suspicious.append(
                    (split, file.name, line_no, width, height)
                )

print("Suspicious tiny polygons:", len(suspicious))

for item in suspicious[:20]:
    print(item)
```

```python
healthy_hashes = []

for img in healthy_images:
    try:
        with Image.open(img) as im:
            im = im.convert("RGB")
            h = imagehash.phash(im)

        healthy_hashes.append({
            "path": img,
            "name": img.name,
            "hash": h
        })

    except Exception as e:
        print("ERROR:", img.name, e)

print("Healthy images processed:", len(healthy_hashes))
```

```python
similar_pairs = []

for i in range(len(healthy_hashes)):
    for j in range(i + 1, len(healthy_hashes)):

        h1 = healthy_hashes[i]["hash"]
        h2 = healthy_hashes[j]["hash"]

        distance = h1 - h2

        if distance <= 8:
            similar_pairs.append({
                "distance": distance,
                "image1": healthy_hashes[i]["name"],
                "image2": healthy_hashes[j]["name"]
            })

print("Similar Healthy image pairs:", len(similar_pairs))

for pair in sorted(similar_pairs, key=lambda x: x["distance"]):
    print("\nDistance:", pair["distance"])
    print("A:", pair["image1"])
    print("B:", pair["image2"])
```

```python
# CODE 1 — Build similarity connections

THRESHOLD = 8

similarity_graph = {
    item["name"]: set()
    for item in healthy_hashes
}

for i in range(len(healthy_hashes)):
    for j in range(i + 1, len(healthy_hashes)):

        h1 = healthy_hashes[i]["hash"]
        h2 = healthy_hashes[j]["hash"]

        distance = h1 - h2

        if distance <= THRESHOLD:
            name1 = healthy_hashes[i]["name"]
            name2 = healthy_hashes[j]["name"]

            similarity_graph[name1].add(name2)
            similarity_graph[name2].add(name1)

print("Images:", len(similarity_graph))
print("Similarity threshold:", THRESHOLD)
```

```python
# CODE 2 — Find connected similarity groups

visited = set()
groups = []

for image_name in similarity_graph:

    if image_name in visited:
        continue

    group = []
    stack = [image_name]

    while stack:

        current = stack.pop()

        if current in visited:
            continue

        visited.add(current)
        group.append(current)

        for neighbor in similarity_graph[current]:
            if neighbor not in visited:
                stack.append(neighbor)

    groups.append(sorted(group))


print("Total Healthy images:", len(healthy_hashes))
print("Possible source groups:", len(groups))

print("\nGroup size distribution:")

from collections import Counter

distribution = Counter(len(group) for group in groups)

for size, count in sorted(distribution.items()):
    print(f"{size} image(s): {count} group(s)")
```

```python
# CODE 3 — Display all Healthy similarity groups

for idx, group in enumerate(
    sorted(groups, key=lambda x: (-len(x), x[0])),
    start=1
):

    print("\n" + "=" * 80)
    print(f"GROUP {idx}")
    print(f"Images in group: {len(group)}")
    print("=" * 80)

    for name in group:
        print(" ", name)
```

```python
# CODE 4 — Show only multi-image groups

multi_groups = [
    group for group in groups
    if len(group) > 1
]

print("Multi-image groups:", len(multi_groups))
print(
    "Images belonging to multi-image groups:",
    sum(len(group) for group in multi_groups)
)

for idx, group in enumerate(
    sorted(multi_groups, key=lambda x: (-len(x), x[0])),
    start=1
):

    print("\n" + "-" * 80)
    print(f"MULTI-IMAGE GROUP {idx} | {len(group)} images")
    print("-" * 80)

    for name in group:
        print(name)
```

```python
# CODE 5 — Compact group summary

print("=" * 80)
print("HEALTHY SOURCE-GROUP SUMMARY")
print("=" * 80)

for idx, group in enumerate(
    sorted(groups, key=lambda x: (-len(x), x[0])),
    start=1
):

    first_name = group[0]

    # Remove augmentation-related suffixes for easier reading
    source_hint = first_name

    for marker in [
        "_rotated_",
        "_color_",
        "_contrast_",
        "_brightness_",
        "_flipped_",
        "_sharpness_",
        "_zoom_",
        "_jpeg"
    ]:
        if marker in source_hint:
            source_hint = source_hint.split(marker)[0]
            break

    print(
        f"Group {idx:02d}: "
        f"{len(group):2d} image(s) | "
        f"Source hint: {source_hint}"
    )
```

```python
# CODE 6 — Inspect augmentation types inside each group

augmentation_markers = [
    "rotated",
    "color",
    "contrast",
    "brightness",
    "flipped",
    "sharpness",
    "zoom",
    "jpeg"
]

for idx, group in enumerate(
    sorted(multi_groups, key=lambda x: (-len(x), x[0])),
    start=1
):

    detected = []

    for name in group:
        name_lower = name.lower()

        for marker in augmentation_markers:
            if marker in name_lower:
                detected.append(marker)
                break

    print(
        f"Group {idx:02d}: "
        f"{len(group)} images | "
        f"augmentation types: {sorted(set(detected))}"
    )
```
